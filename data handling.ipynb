```python
# transformer- performs preprocessing eg. OneHotCoding, StandardScaler
#pipeline- puts preprocessing + ML model into one sequence 
#column_transformer- tells sklearn which columns nees which transformation
```

```python
from sklearn.compose import ColumnTransformer
```

```python
from sklearn.pipeline import Pipeline
```

```python
from sklearn.preprocessing import OneHotEncoder , StandardScaler
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
categorical_features = ['sex','cp','fbs','restecg','exang','slope','ca','thal']
```

```python
numerical_features = ['age','trestbps','chol','thalach','oldpeak']
```

```python
preprocessor = ColumnTransformer(transformers =[('num',StandardScaler(),numerical_features),('cat',OneHotEncoder(handle_unknown = 'ignore'),categorical_features)])
```

```python
model =Pipeline([
    ('preprocessor',preprocessor),
    ('classifier',LogisticRegression())])
```

```python
model.fit(x_train,y_train)
```

```python
y_pred  = model.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score
```

```python
print(accuracy_score(y_test ,y_pred))
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
model =Pipeline([
    ('preprocessor',preprocessor),
    ('classifier',RandomForestClassifier())])
```

```python
model.fit(x_train,y_train)
```

```python
y_pred  = model.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score
```

```python
print(accuracy_score(y_test ,y_pred))
```
